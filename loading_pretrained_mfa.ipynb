# Load a Pretrained MFA

Minimal example for loading an MFA checkpoint and moving it to your target device.

## Imports

```python
import torch
from huggingface_hub import hf_hub_download
from modeling.model_checkpointing import load_mfa
```

### Loading Pre-trained

This notebook loads checkpoints using `load_mfa` from `modeling.model_checkpointing`.

The main MFA guide can load from `modeling.mfa`, but that path is for checkpoints saved with the model-level saver.

When loading the published pretrained MFAs Use the checkpointing loader shown here because these checkpoints were saved with Accelerate/FSDP and include checkpoint-specific metadata/artifacts.

## Available Pretrained MFAs (8k)

The currently released checkpoints are the **8k MFAs** from the paper, trained on the residual stream.

- **Gemma-2-2B**: layers **6** and **18**
- **Llama-3.1-8B**: layers **8** and **22**


These MFAs were trained on a **subset of The Pile**, so they cover a specific region of activation space rather than the full distribution.

More pretrained MFA checkpoints will be released soon with varying sizes.

## Download Checkpoint from Hugging Face

Set your repo and filename, then run this once to cache the checkpoint locally.

```python
# For Gemma-2-2b, use the following repo_id:
# repo_id = "ordavids1/MFA-8k-gemma-2-2b"

# For Llama-3.1-8B, use the following repo_id:
repo_id = "ordavids1/decomposing-local-geometry-MFAs"

# For Gemma-2-2b, we have layers 18 and 6 and for Llama-3.1-8B, we have layers 22 and 8. Make sure to update the filename accordingly -- look at huggingface for exact names.
filename = "gemma-2-2b_mfa_l18_8k.ckpt"  # e.g., "checkpoints/mfa.pt" if stored in a subfolder
repo_type = "model"

ckpt_local_path = hf_hub_download(
    repo_id=repo_id,
    filename=filename,
    repo_type=repo_type,
 )

print("Downloaded to:", ckpt_local_path)
```

## Load Checkpoint

```python
checkpoint_path = ckpt_local_path if "ckpt_local_path" in globals() else "./checkpoints/mfa.pt"
device = "mps" if torch.backends.mps.is_available() else ("cuda" if torch.cuda.is_available() else "cpu")

mfa = load_mfa(checkpoint_path, map_location=device)
mfa.eval()

print(f"Loaded MFA on {device}")
print(f"Checkpoint: {checkpoint_path}")
print(f"K={mfa.K}, D={mfa.D}, q={mfa.q}")
```

### Usage

All utils (steering/visualization/etc.) shown in the main tutorial will work as shown on the loaded model.
